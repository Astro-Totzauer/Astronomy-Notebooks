```python
import os
import glob

def delete_files_with_phrase(directory, phrase):
    """Deletes files in the specified directory containing the given phrase."""

    for filename in glob.glob(os.path.join(directory, "*")):
        if phrase in filename:
            try:
                os.remove(filename)
                print(f"Deleted: {filename}")
            except OSError as e:
                print(f"Error deleting {filename}: {e}")

if __name__ == "__main__":
    directory_to_search = os.getcwd() 
    phrase_to_match = ":Zone.Identifier"

    delete_files_with_phrase(directory_to_search, phrase_to_match)
```
